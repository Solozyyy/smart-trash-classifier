## Sử dụng config trong code

````python
import os

# Lấy các biến từ config
DATA_DIR = config.DATA_DIR
BALANCED_DIR = config.BALANCED_DIR
TARGET_COUNT = config.TARGET_COUNT
BATCH_SIZE = config.BATCH_SIZE

print(f" Data directory: {DATA_DIR}")
print(f" Balanced directory: {BALANCED_DIR}")
print(f" Target images per class: {TARGET_COUNT}")
print(f" Batch size: {BATCH_SIZE}")

# Kiểm tra xem data directory có tồn tại không
if os.path.exists(DATA_DIR):
    categories = [d for d in os.listdir(DATA_DIR) if os.path.isdir(os.path.join(DATA_DIR, d))]
    print(f"\n Tìm thấy {len(categories)} categories: {categories}")
else:
    print(f"\n  Không tìm thấy data directory tại {DATA_DIR}")
````

## Example: Load và hiển thị ảnh

````python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import random

# Lấy 5 ảnh ngẫu nhiên từ data
if os.path.exists(DATA_DIR):
    categories = [d for d in os.listdir(DATA_DIR) if os.path.isdir(os.path.join(DATA_DIR, d))]
    
    plt.figure(figsize=(20, 4))
    for i in range(5):
        random_cat = random.choice(categories)
        cat_path = os.path.join(DATA_DIR, random_cat)
        random_img = random.choice(os.listdir(cat_path))
        img_path = os.path.join(cat_path, random_img)
        
        img = mpimg.imread(img_path)
        plt.subplot(1, 5, i + 1)
        plt.imshow(img)
        plt.title(random_cat)
        plt.axis('off')
    
    plt.tight_layout()
    plt.show()
else:
    print("  Data directory không tồn tại. Vui lòng kiểm tra config.")
````
